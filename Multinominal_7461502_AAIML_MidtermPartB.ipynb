# Wine Classification Using Naive Bayes Classifiers

## Project Description

The objective of this project is to develop and evaluate machine learning models to classify wine samples into three different categories using the Wine dataset available in `sklearn.datasets`.

In this project, two Naive Bayes classification algorithms, **Gaussian Naive Bayes** and **Multinomial Naive Bayes**, are trained and tested to predict the category of each wine sample based on its chemical properties. The dataset is divided into training and testing sets to evaluate the performance of both models on unseen data.

The models are compared using accuracy scores, classification reports, and confusion matrices to determine which classifier performs better. Finally, predictions are made on selected test samples and compared with their actual categories.

The project aims to identify which of the two classifiers achieves better classification performance on the Wine dataset and to understand how the choice of algorithm and its underlying assumptions influence prediction results.

```python

import pandas as pd

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

### 1. Load Data Set

```python

# Load the Wine dataset
wine = load_wine()

# Separate the input features and target labels
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = wine.target

# Display the first five rows
display(X.head())

# Display dataset information
print("Dataset shape:", X.shape)
print("Number of samples:", len(y))
print("Number of features:", X.shape[1])
print("Wine classes:", wine.target_names)

# Count the samples in each class
print("\nSamples per class:")
print(pd.Series(y).value_counts().sort_index())
```


### Observation: Dataset Exploration

The Wine dataset contains 178 samples and 13 numerical features describing the chemical properties of wine. The target variable consists of three classes: 0, 1, and 2. These features will be used to train machine learning models to classify the wine samples.


### 2.Split the dataset into training and testing sets
Use 80% of the data to train the models and 20% to test their performance on unseen samples.

```python

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the sizes of the resulting datasets
print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```


### Observation: Train-Test Split

The dataset was divided into training and testing sets using an 80:20 ratio. The training set contains 142 samples, while the testing set contains 36 samples. Stratified sampling was used to maintain approximately the same class distribution in both sets. The training data will be used to fit the classifiers, and the testing data will be used to evaluate their performance on unseen samples.


### 3. Train the Gaussian Naive Bayes classifier

```python

# Create the Gaussian Naive Bayes classifier
gaussian_model = GaussianNB()

# Train the model using the training data
gaussian_model.fit(X_train, y_train)

# Predict wine categories using the test data
gaussian_predictions = gaussian_model.predict(X_test)

# Calculate the model's accuracy
gaussian_accuracy = accuracy_score(y_test, gaussian_predictions)

print("Gaussian Naive Bayes Model")
print("--------------------------")
print("Accuracy:", round(gaussian_accuracy * 100, 2), "%")
print("Correct predictions:", (gaussian_predictions == y_test).sum())
print("Total test samples:", len(y_test))
```


### Observation: Gaussian Naive Bayes

The Gaussian Naive Bayes classifier was trained using the training dataset and evaluated using the test dataset. The model predicted the wine categories based on the chemical features. Its accuracy indicates how well it classified previously unseen samples.


### 4. Train the Multinomial Naive Bayes classifier
Now let's train the second model using the same training and testing sets so that we can compare the results fairly.

```python

# Check that all features are non-negative
print("Minimum training feature value:", X_train.min().min())
print("Minimum testing feature value:", X_test.min().min())

assert (X_train >= 0).all().all(), "Training data contains negative values"
assert (X_test >= 0).all().all(), "Testing data contains negative values"

print("All features are non-negative.")

# Create the Multinomial Naive Bayes classifier
multinomial_model = MultinomialNB()

# Train the model using the training data
multinomial_model.fit(X_train, y_train)

# Predict wine categories using the test data
multinomial_predictions = multinomial_model.predict(X_test)

# Calculate the model's accuracy
multinomial_accuracy = accuracy_score(
    y_test, multinomial_predictions
)

print("Multinomial Naive Bayes Model")
print("-----------------------------")
print("Accuracy:", round(multinomial_accuracy * 100, 2), "%")
print("Correct predictions:",
      (multinomial_predictions == y_test).sum())
print("Total test samples:", len(y_test))
```

### 5. Compare both models

```python

# Create a comparison table
results_df = pd.DataFrame({
    "Model": [
        "Gaussian Naive Bayes",
        "Multinomial Naive Bayes"
    ],
    "Accuracy (%)": [
        gaussian_accuracy * 100,
        multinomial_accuracy * 100
    ],
    "Correct Predictions": [
        (gaussian_predictions == y_test).sum(),
        (multinomial_predictions == y_test).sum()
    ],
    "Incorrect Predictions": [
        (gaussian_predictions != y_test).sum(),
        (multinomial_predictions != y_test).sum()
    ]
})

# Sort models by accuracy
results_df = results_df.sort_values(
    by="Accuracy (%)",
    ascending=False
).reset_index(drop=True)

display(results_df)

# Identify the best-performing model
best_model_name = results_df.loc[0, "Model"]

print("Best-performing model:", best_model_name)
```

### 6. Evaluate predictions using classification reports

```python

print("GAUSSIAN NAIVE BAYES")
print(classification_report(
    y_test,
    gaussian_predictions,
    target_names=wine.target_names
))

print("MULTINOMIAL NAIVE BAYES")
print(classification_report(
    y_test,
    multinomial_predictions,
    target_names=wine.target_names
))
```

### 7. Make predictions on five test samples

```python


# Select five samples from the test dataset
sample_data = X_test.iloc[:5]

# Predict using both trained models
gaussian_sample_predictions = gaussian_model.predict(sample_data)
multinomial_sample_predictions = multinomial_model.predict(sample_data)

# Create a comparison table
prediction_results = pd.DataFrame({
    "Test Sample Index": sample_data.index,
    "Actual Category": wine.target_names[y_test[:5]],
    "Gaussian Prediction": wine.target_names[gaussian_sample_predictions],
    "Multinomial Prediction": wine.target_names[multinomial_sample_predictions]
})

# Display the results
display(prediction_results)

```


### Observation: Test Data Predictions

Both trained classifiers correctly predicted the wine categories for the five selected test samples. The predicted categories matched the actual categories for all five samples.

Although both models performed equally well on these selected samples, their overall test accuracies were different. Gaussian Naive Bayes achieved 97.22% accuracy, while Multinomial Naive Bayes achieved 88.89%. This demonstrates that evaluating multiple test samples is useful, but overall model performance should be assessed using the complete test dataset.
